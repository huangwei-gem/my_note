## 1.1Numpy 简介

- 一个个基于python的扩展库
- 提供高维数组对象ndarray，运算速度碾压python List
- 提供了各种高级数据编程工具，如矩阵运算、向量运算、快速筛选、10操作、傅里叶变换、线性代数、随机数等

## 1.2 ndarray属性

1. ndim：维度
2. shape：形状(各维度的长度)
3. size：总长度
4. dtype：元素类型
