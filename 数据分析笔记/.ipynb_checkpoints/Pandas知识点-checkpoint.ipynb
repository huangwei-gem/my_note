# 1 Pandas 核心数据类型
## 1.1 什么是pandas

[Pandas官方文档](https://pandas.ac.cn/docs/user_guide/index.html)
- Pandas是Python语言的一个扩展程序库，用于数据分析。
- Pandas是一个开放源码、BSD许可的库，提供高性能、易于使用的数据结构和数据分析工具。
- Pandas名字衍生自术语“panel data”（面板数据）和"Python data analysis"（Python数据分析）。
- Pandas一个强大的分析结构化数据的工具集，基础是Numpy（提供高性能的矩阵运算）。
- Pandas可以从各种文件格式比如CSV、JSON、SQL、MicrosoftExcel导入数据。
- Pandas可以对各种数据进行运算操作，比如归并、再成形、选择，还有数据清洗和数据加工特征。


Pandas的主要数据结构是Series（一维数据）与DataFrame（二维数据），这两种数据结构足以处理金融工程等领域里的大多数典型用例

## 1.2 Series
Series是一种类似于字典（dict-like）的一维数组（array-like）的对象，由下面俩个部分组成：
- values：一组数据（ndarray类型）
- index：相关的数据索引标签
  

```python
import numpy as np
import pandas as pd
```

### 1.2.1 构造Series
1. from ndarray使用ndarray构造是一个引用对象（其实就是引用了地址）修改对应的值回修改原数据
   

```python
arr1 = np.random.randint(0,10,size=5)
arr1
```

```python
pd.Series(data=arr1)
```

```python
# index 用来指定Series对象的显示索引，如果没有显示设置索引，那就是默认索引0,1,2,3,4,5
pd.Series(data=arr1,index=['A','B','C','D','E'])
```

2. from list使用列表构造 是一副本对象(复制了一份到一块储存）不会修改原数据

```python
s1='ABCD'
list(s1)
```

3. from dict 使用字典构造

```python
user_info = {
    'name':'Richa',
    'age':18,
    'address':'北京'
}

user_info
```

```python
# 字典的key会作为索引，values会作为内容
s1 = pd.Series(data=user_info)
s1
```

```python
# NaN not a number 空数值
s2 = pd.Series(data=user_info,index=list('ABC'))
s2
```

```python
# 对于dict构造，index优先级更高，指定的index内部，如果没有就为空值（NaN）
pd.Series(data=user_info,index=['A','B','C','name','age','address'])
```

4. From scalar value
- 使用一个标量（常量或变量）构造时，必须指定index，标量会被重复匹配u到每一个index

```python
l2 = [1,2,3,4,5,6]
pd.Series(data=l2,index=list('ABCDEF'))
```

```python
# 广播机制
pd.Series(data=1,index=list('ABCDE'))
```

### 1.2.3 Series的数学运算
Series对大多数NumPy的函数都是兼容的
1. 与非pandas对象运算
服从广播机制原则

```python
s = pd.Series(data=[1,2,3],index=list("ABC"))
s
```

```python
s+1
```

```python
arr1 = np.array([1,2,3])
arr1,s+arr1
```

2. Numpy functions

```python
arr1,np.power(arr1,2)
```

```python
# ndarray对象和pandas对象都支持Numpy的函数操作
s,np.power(s,2)
```

3. Series之间运算
- 索引对齐原则
- 对不起补空值，使用add\sub\mul\div函数处理空值


```python
s1=pd.Series(data=np.array([4,5,6]),index=list("CBA"))
s1,s,s1+s                
```

## 1.3 DataFrame
DataFrame是一个二维的表格型数据结构，可以看做是由Series组成的字典（共用同一个索引l）。

DataFrame由按一定顺字排列的【多列】数据组成，每一列的数据类型可能不同

设计初衷是将Series的使用场景从一维拓展到多维。DataFrame既有行索引，也有列索引。
- 行索引l：index
- 列索引l：columns
- 值：values（numpy的二维数组）

```python
import numpy as np
import pandas as pd
```

```python
# serieshe ndarray一样会强制类型统一
list1 = [1,2,'Richa']
s1=pd.Series(data=list1)
s1
```

dataframe注意事项⚠️

给我的感觉就是和exel一样就行

- 1.DataFrame而言，列与列之间的数据类型可以不一样，但是对于单个列（Series）而言，数据类型强制统一
- 2.columns概念，列名，（Series的name属性），通过DataFrame调用一个列名的时候，会输出整个Series的value
- 3.由多个Series组成的，共用index

### 1.3.1 构造dataframe
1. from dict of series or dicts 使用一个由series构造的字典或一个字典构造
2. from dict of ndarrays/lists 使用一个由列表或ndarray构造的字典结构
3. from a list ofdicts 使用一个由字典构成的列表构造
4. dataframe.from_dict()函数构造

```python
data = np.random.randint(0,100,size=(3,4))
index = ['Tom','Lucy','Dancer']
columns = ['语文','数学','英语','化学']
pd.DataFrame(data,index,columns)
```

```python

names = pd.Series(data=['Tom','Lucy','Dancer'],index=list('ABC'))
scores = pd.Series(data=np.random.randint(0,100,size=3),index=list('ABC'))
pd.DataFrame(data={
    'name':names,
'scores':scores
})
```

```python
dic=[{
    'name':'Lucy',
    'score':99
},{
    'name':'Tim',
'score':29
}]
pd.DataFrame(data=dic,index=[0,1])
```

```python
names=['Tom','Lucy','Dancer']
chinese=np.random.randint(1,100,size=3)
math=np.random.randint(0,100,size=3)
display(names,chinese,math)
```

```python
student_dict={
    'names':names,
    '语文':chinese,
    '数学':math
}
score = pd.DataFrame(data=student_dict)
```
