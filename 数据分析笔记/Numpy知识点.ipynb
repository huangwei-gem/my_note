### 1.6.2广播Broadcast
ndarray广播机制的俩条规则
如果
1. 俩个数组的后缘维度（即从末尾开始算起的维度）的轴长度相符
2. 或者其中的一方的长度为1
则认为它们是广播兼容。广播会在缺失和（或）长度为1的维度上进行。

例1：a.np.ones((2,3)) b.np.arange(3) 求a+b

例2：a=np.ones(4,3,2) b=np.random.randint(0,10,size(3,2)),求a+b

例3 ：a=np.arange(3),reshape((3,1)),b=np.arange(3).求a+b

```python
def standardize_by_std(data):
    """
    标准Z-score标准化：减去均值，再除以样本标准差（实际应用中更常用）
    
    参数：
        data: 待标准化的数据，可以是Python列表或numpy ndarray
    返回：
        numpy ndarray: 标准化后的数组
    """
    arr = np.array(data, dtype=np.float64)
    mean_val = np.mean(arr)
    sample_std = np.std(arr, ddof=1)  # 样本标准差（样本方差的平方根）
    
    if sample_std == 0:
        raise ValueError("样本标准差为0，无法完成标准化（所有数据值相同）")
    
    standardized_arr = (arr - mean_val) / sample_std
    return standardized_arr
```
