## 1.6习题
背景：假设班级有张三、李四、王五、赵六四位同学，考试要考的科目有Python、Java、C
1. 假设score1是期中考试成绩，score2是期末考试成绩，请自由创建score1和score2，并将其相加，求期中期末平均值。
2. 假设张三期中考试Java被发现作弊，要记为0分，如何实现？
3. 李四因为举报张三作弊立功，期中考试所有科目加10分，如何实现？
4. 由于有一道题出错，要给所有学生加10分，如何实现？
5. Python老师想知道哪些同学的Java成绩比Python好，如何实现？


```python
import pandas as pd
import numpy as np
```

```python
score1 = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    'Python': np.random.randint(0, 101, size=4),
    'Java': np.random.randint(0, 101, size=4),
    'C': np.random.randint(0, 101, size=4)
})
score2 = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    'Python':np.random.randint(0, 101, size=4),
    'Java': np.random.randint(0, 101, size=4),
    'C': np.random.randint(0, 101, size=4)
})
display(score1)
display(score2)
```

```python
result = (score1 + score2).mean(axis=1, numeric_only=True)
display(result)
```

```python
score1.loc['张三','Java']=0
score1
```

```python
# 1. 筛选出数值列和非数值列
numeric_cols = score1.select_dtypes(include='number')  # 数值列（如Python、Java、C）
non_numeric_cols = score1.select_dtypes(exclude='number')  # 非数值列（如姓名）

# 2. 仅对数值列执行加法
numeric_result = numeric_cols.add(10)  # 或 numeric_cols + 10

# 3. 合并数值列结果和非数值列（保持原顺序）
result = pd.concat([non_numeric_cols, numeric_result], axis=1)[score1.columns]
result
```
