```python
data.groupby('V1')['V2'].mean()
```

## 1.4 交叉表
统计的是频次

```python
pd.crosstab(data['V1'],data['V2'])
```

## 1.5 透视表
透视表和交叉表的区别就是，透视表不只是统计频次，还可以统计其他的指标，比如均值、中位数、标准差等。

```python
pd.pivot_table(data=data,index='V1',columns='V2',values='V3',aggfunc=np.mean)
```

## 1.6 查找（isin）

## 1.7 区间划分（cut）

```python
data['V3'].map(lambda x: '高' if x>100 else '低')
```

```python
bins = [data['V2'].min()-1,10,20,30,40,50,data['V2'].max()]
pd.cut(data['V2'],bins=bins,right=False)
```
