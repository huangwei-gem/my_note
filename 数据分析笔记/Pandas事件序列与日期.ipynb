## 1.2.3提供格式参数format
- 要实现精准转换，除了传递datetime字符串，还可以通过format参数来指定格式，指定此参数还可以加速转换速度。
- Y年m月d日H时M分S秒

```python


from datetime import datetime

# 待解析的 datetime 字符串
datetime_str = "2026年01月23日15时30分45秒"

# 定义对应的格式（与字符串格式完全匹配）
format_pattern = "%Y年%m月%d日%H时%M分%S秒"

# 使用 strptime + format 参数进行精准转换
dt = datetime.strptime(datetime_str, format_pattern)

print(dt)  # 输出：2026-01-23 15:30:45
```

### 1.2.4无效数据
- 不可解析时，默认值errors='raise'会触发错误：
- errors=ignore'返回原始输入：
- errors='coerce'把无法解析的数据转换为NaT，即不是时间（NotaTime）：

```python
# 2012-12-31
# 2012年12月31号
# None
# 2012/12/12
# 上面的那种类型不是平常的类型。所以说，在转换的时候，需要指定errors='ignore'或者其他的参数。
pd.to_datetime(['2012/12/12','2012年12月31号',None,'asd'],errors='ignore')
```

```python
# 使用率最高的
pd.to_datetime(['2012/12/12','2012年12月31号',None,'asd'],errors='coerce')
```

```python
def my_transform(x):
    if x is None:
        return None
    if x.endswith('号'):
        return pd.to_datetime(x.replace('号','日'),format='%Y年%m月%d日')
    else:
        return x
```

```python
pd.Series(['2012/12/12','2012年12月31号',None,'asd']).map(my_transform)
```

## 1.6重采样（每几年做求和）
- Pandas有一个虽然简单，但却强大、高效的功能，可在频率转换时执行重采样，如，将秒数据转换为5分钟数据，这种操作在金融等领域里的应用非常广泛。
- resample()是基于时间的分组操作，每个组都遵循归纳方法。
- 从0.18.0版开始,resample()可以直接用于DataFrameGroupBy对象
> 注意:重采样一定要以时间类型做Index
- resample函数非常灵活，可以指定多种频率转换与重采样参数。
- 支持聚合函数的处理如:sum、mean、std、sem、max、min、mid、median

```python
AAPL1 = AAPL.set_index('register_time')
AAPL1
```

```python
AAPL['register_year'] = AAPL['register_time'].dt.year
AAPL
```

```python
# 每三个月对学习时间（learn_time）求平均值
AAPL1.resample('3M')['learn_time'].mean()
```
