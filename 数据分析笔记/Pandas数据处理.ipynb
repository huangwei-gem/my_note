
## 1.4异常值处理

 体育比赛里，最后算分的时候总会去掉一个最高分，去掉一个最低分，将剩下的分数进行去平均。因为一个非常夸张的异常值可能会造成对最后统计结果产生比较大的影响。所以，异常值处理，在统计数据的过程中是必然要做的一个操作
### 1.4.1 异常值分析
- 使用describe()方法可以查看数据的基本统计信息，包括均值、标准差、最小值、25%、50%、75%、最大值等。
- order = pd.read_csv('canyin2019.csv')
- order.describe()

### 1.4.2 异常值的认定
- 不符合业务逻辑的异常值
- 使用均值和标准差进行判断
- 使用上四中位数和下四中位数进行异常值判定

```python
# 数据要先满足服从正态分布，才能使用均值和标准差进行异常值判断
df = pd.DataFrame(data=np.random.randn(1000,4),columns=list('ABCD'))
df
```

```python
import seaborn as sns
sns.histplot(df.A)
```

```python
df.loc[~(np.abs(df)>(3*df.std())).any(axis=1)]
```

```python
df.drop(index=df.loc[(np.abs(df)>(3*df.std())).any(axis=1)].index)
```

```python
order = pd.read_csv('canyin2019.csv')
order.describe([.99]).T
```

### 1.4.3 异常值的处理

```python
# 一种数学的异常值判定方法
Q1 = order['V2'].quantile(0.25)
Q3 = order['V2'].quantile(0.75)
IQR = Q3 - Q1
Top = Q3 + 1.5*IQR
Bottom = Q1 - 1.5*IQR
display(Top,Bottom)
```

```python
len(order[(order['V2']>Top)|(order['V2']<Bottom)])
len(order)
display(len(order[(order['V2']>Top)|(order['V2']<Bottom)]),len(order))
```

```python
# 异常值的处理(可以给平均值，也可以给中位数，也可以给最高值或者最低值，主要是取决于业务的导向)
# 这里是取最大值
order[(order['V2']>Top)|(order['V2']<Bottom)]=Top
order
```

```python
import datetime
import numpy as np
import pandas as pd
df = pd.read_csv('canyin2019.csv')
# 优势：节省人力成本，节约繁琐的操作，容易出错
start = 0
for i in range(int(np.ceil(len(df)/99))):
    df.iloc[start:start+99].to_excel(f'{i}.xlsx',index=False)
    start+=99
    print(f'第{i}个文件已保存')


```
