## 1.1 获取画板

```python
import matplotlib.pyplot as plt
# 要把当前要回执的图形加载到画布之上
%matplotlib inline 
```

```python
# 画板是不呈像的，画板只负责管理画布
plt.figure(figsize=(10,6))
```

## 1.2 获取画布

```python
fig = plt.figure(figsize=(10,6))
# 子画布
# 1行2列 第2个
plt.subplot(1,2,2)
plt.subplot(2,2,1)
```



## 1.3 就近原则
- plt.的所有操作都是操作最近的上一个画布
- 所有的画布，都会默认被配置在离自己最近的那个画板上

```python
import numpy as np
import pandas as pd
```

```python
x = np.linspace(0,2*np.pi,100)
y = np.sin(x)
# 绘制线性图
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,y)
```

## 1.4 网格线的构造

```python
# 也可以面向对象来创建
sub1 = plt.subplot(2,2,1)
sub1.plot(x,y)
sub2 = plt.subplot(1,2,2)
sub2.plot(x,y)
```

## 1.5 坐标轴刻度界限


```python
# 默认情况下，Matplotlib会自动设置坐标轴的刻度界限，以确保数据能够完整显示。
plt.plot(x,y)
```

```python
plt.plot(x,y)
plt.axis([0,10,-2,2])
```

## 1.6 坐标轴刻度标签


```python
plt.plot(x,y)
plt.axis([0,10,-2,2])
plt.xticks([0,2,4,6,8,10])
plt.yticks([-2,-1,0,1,2])
plt.xlabel('x')
plt.ylabel('y')
plt.title('y=sin(x)')
```

## 1.7 轴标签和画布标题


```python
plt.plot(x,y)
plt.axis([0,10,-2,2])
plt.xticks([0,2,4,6,8,10])
plt.yticks([-2,-1,0,1,2])
plt.xlabel('x')
plt.ylabel('y')
plt.title('y=sin(x)')
```

## 1.10 图例
### 1.10.1 俩种添加图例的方法
- 直接在plot函数中添加label参数
- 调用legend函数

```python
plt.plot(x,y,label='y=sin(x)')
plt.plot(x,np.cos(x),label='y=cos(x)')
plt.legend()
```

```python
plt.plot(x,y)
plt.plot(x,np.cos(x))
plt.legend(['y=sin(x)','y=cos(x)'])
```

### 1.10.2 loc参数
- loc参数用于指定图例的位置
- 常用的位置参数有：'upper right'、'upper left'、'lower left'、'lower right'
- 也可以使用数字表示位置，如：0表示'upper right'，1表示'upper left'，2表示'lower left'，3表示'lower right'

```python
plt.figure(figsize=(10,6))
plt.plot(x,y)
plt.plot(x,np.cos(x))
plt.legend(['y=sin(x)','y=cos(x)'],loc='upper right')
```

## 1.11 图片保存


```python
figure = plt.figure(figsize=(10,6))
axes = plt.subplot(1,1,1)
plt.plot(x,y)
plt.plot(x,np.cos(x))
plt.legend(['y=sin(x)','y=cos(x)'],loc='upper right')
plt.title('y=sin(x) and y=cos(x)')
plt.xlabel('x')
plt.ylabel('y')
# dpi 分辨率
plt.savefig('sin_cos.png',dpi=500)
```

# 2 添加注解
添加文字的方法：
- plt.text(x,y,'text')
- plt.annotate('text',xy=(x,y),xytext=(x,y),arrowprops=dict(facecolor='red',shrink=0.05))
- plt.figtext(x,y,'text')
- plt.suptitle('title')


```python
#x,y表示文字在画布中的位置信息，使用画布刻度值来描述的
figure = plt.figure(figsize=(10,6))
ax1 = plt.subplot(1,1,1)
plt.plot(x,y)
plt.text(0.5,0.5,'y=sin(x)')
ax2=plt.subplot(1,2,2)
plt.plot(x,np.cos(x))
plt.text(0.5,np.cos(0.5),'y=cos(x)')
# 为整个画布添加标题,使用相对位置
figure.suptitle('y=sin(x) and y=cos(x)')

```


## 2.1 图形内添加文字

```python
figure = plt.figure(figsize=(10,6))
ax1 = plt.subplot(1,1,1)
plt.plot(x,y)
plt.text(0.5,0.5,'y=sin(x)')
ax2=plt.subplot(1,2,2)
plt.plot(x,np.cos(x))
plt.text(0.5,np.cos(0.5),'y=cos(x)')
# 箭头指向最高点
ax2.annotate('max',xy=(0,np.cos(0)),xytext=(1,np.cos(1)),arrowprops=dict(facecolor='red',shrink=0.05))
# 为整个画布添加标题,使用相对位置
figure.suptitle('y=sin(x) and y=cos(x)')
# 为每个子图添加标题
ax1.set_title('y=sin(x)')
ax2.set_title('y=cos(x)')
```
